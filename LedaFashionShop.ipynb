#### SRT Transcript Generator

````python
import os
import json
from google import genai
from google.genai import types
from dotenv import load_dotenv

load_dotenv()
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")

def generate_srt_from_audio(audio_file_path, output_srt_path="output.srt"):
    """
    Uploads an audio file to Gemini, requests a timed transcript,
    and saves the result as an SRT subtitle file.
    """
    # 1. Initialize the GenAI Client
    client = genai.Client()

    # 2. Upload the audio file using the Files API
    print(f"Uploading {audio_file_path} to Gemini...")
    audio_file = client.files.upload(file=audio_file_path)
    print(f"Upload complete: {audio_file.name}")

    # 3. Define the Prompt
    prompt = """
    Process this audio file and generate a highly accurate transcription suitable for subtitles.
    Requirements:
    1. Break the transcription into short, logical segments (e.g., 1-2 sentences per segment).
    2. Provide accurate start and end timestamps for each segment in the exact SRT time format: HH:MM:SS,mmm (e.g., 00:01:23,450).
    3. Provide the exact transcribed text for that segment.
    """

    # 4. Enforce Structured JSON Output (Schema)
    print("Analyzing audio and generating timestamps... This may take a moment.")
    response = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=[
            types.Part.from_text(text=prompt),
            audio_file
        ],
        config=types.GenerateContentConfig(
            # FIXED: Use response_mime_type instead of the nested response_format dict
            response_mime_type="application/json",
            response_schema=types.Schema(
                type=types.Type.OBJECT,
                properties={
                    "segments": types.Schema(
                        type=types.Type.ARRAY,
                        description="List of transcribed segments with SRT timestamps.",
                        items=types.Schema(
                            type=types.Type.OBJECT,
                            properties={
                                "start_time": types.Schema(
                                    type=types.Type.STRING,
                                    description="Exact start time in HH:MM:SS,mmm format"
                                ),
                                "end_time": types.Schema(
                                    type=types.Type.STRING,
                                    description="Exact end time in HH:MM:SS,mmm format"
                                ),
                                "text": types.Schema(
                                    type=types.Type.STRING,
                                    description="The spoken text"
                                ),
                            },
                            required=["start_time", "end_time", "text"],
                        ),
                    ),
                },
                required=["segments"],
            ),
        ),
    )

    # 5. Parse the JSON response
    try:
        # Sometimes the API wraps raw JSON in Markdown backticks even with mime_type set.
        # We strip them safely before parsing.
        raw_json = response.text.strip("```json\n").strip("```")
        data = json.loads(raw_json)
        segments = data.get("segments", [])
    except json.JSONDecodeError as e:
        print(f"Error parsing Gemini's output: {e}")
        print("Raw output:", response.text)
        return

    # 6. Build the SRT File
    print(f"Writing subtitles to {output_srt_path}...")
    with open(output_srt_path, "w", encoding="utf-8") as srt_file:
        for index, segment in enumerate(segments, start=1):
            start = segment.get("start_time", "00:00:00,000")
            end = segment.get("end_time", "00:00:00,000")
            text = segment.get("text", "").strip()

            # SRT Format Structure:
            # [Sequence Number]
            # [Start Time] --> [End Time]
            # [Subtitle Text]
            # [Blank Line]
            srt_file.write(f"{index}\n")
            srt_file.write(f"{start} --> {end}\n")
            srt_file.write(f"{text}\n\n")

    print("SRT generation complete!")

if __name__ == "__main__":
    # --- Execute Script ---
    INPUT_AUDIO = "/content/sample_data/audio.wav"
    OUTPUT_FILE = "transcript.srt"

    generate_srt_from_audio(INPUT_AUDIO, OUTPUT_FILE)
````

Using Pydantic for structured output

````python


import os
from google import genai
from pydantic import BaseModel, Field

# Ensure API key is set in environment variables
if not os.getenv("GEMINI_API_KEY"):
    raise ValueError("GEMINI_API_KEY environment variable not set")

# 1. Define a Pydantic model for the desired output structure
class Cat(BaseModel):
    """Information about a cat."""
    name: str = Field(description="The name of the cat.")
    colour: str = Field(description="The color of the cat.")
    special_ability: str = Field(description="A unique ability the cat possesses.")

# 2. Initialize the Gemini client (it automatically picks up the API key)
client = genai.Client()

# 3. Define the prompt
prompt = "Generate data for 3 cats, including their name, colour and special ability."

# 4. Call the API, specifying the response schema
response = client.models.generate_content(
    model="gemini-2.5-flash", # Use a model that supports structured output
    contents=prompt,
    config={
        "response_mime_type": "application/json",
        "response_schema": list[Cat], # Specify a list of Cat objects
    },
)

# 5. The response is automatically parsed into a list of Pydantic objects
my_cats: list[Cat] = response.parsed
# You can also access the raw JSON string with response.text

# 6. Print the generated data in a type-safe manner
for cat in my_cats:
    print(f"Name: {cat.name}, Colour: {cat.colour}, Special Ability: {cat.special_ability}")
````

#### ASS Transcript Generator Node

````python
import os
import json
from google import genai
from google.genai import types
from dotenv import load_dotenv

load_dotenv()
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")

def srt_to_ass_time(srt_time_str):
    """
    Converts SRT time format (HH:MM:SS,mmm) to ASS time format (H:MM:SS.cs).
    ASS uses centiseconds (2 digits) instead of milliseconds (3 digits).
    """
    parts = srt_time_str.strip().replace(',', '.').split(':')
    if len(parts) != 3:
        return "0:00:00.00"

    h = int(parts[0])
    m = parts[1]
    s_ms = float(parts[2])

    # Format to H:MM:SS.cs (e.g., 0:01:23.45)
    return f"{h}:{m}:{s_ms:05.2f}"

def generate_ass_from_audio(audio_file_path, output_ass_path="output.ass"):
    """
    Uploads an audio file to Gemini, requests word-level timestamps,
    and saves the result as an animated ASS subtitle file.
    """
    # 1. Initialize the GenAI Client
    client = genai.Client()

    # 2. Upload the audio file using the Files API
    print(f"Uploading {audio_file_path} to Gemini...")
    audio_file = client.files.upload(file=audio_file_path)
    print(f"Upload complete: {audio_file.name}")

    # 3. Define the Prompt tailored for word-by-word animation
    prompt = """
    Process this audio file and generate an extremely precise, word-by-word transcription.
    Requirements:
    1. Group the audio into short sentence/phrase sequences.
    2. For each sequence, provide the overall start and end times.
    3. Inside each sequence, break the phrase into an ordered list of individual words, specifying the exact start and end times for when that individual word is spoken.
    4. Timestamps must be in the format: HH:MM:SS,mmm
    """

    # 4. Enforce Structured JSON Output (Schema containing nested words)
    print("Analyzing audio and generating word-level timestamps... This may take a moment.")
    response = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=[
            types.Part.from_text(text=prompt),
            audio_file
        ],
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=types.Schema(
                type=types.Type.OBJECT,
                properties={
                    "sequences": types.Schema(
                        type=types.Type.ARRAY,
                        description="List of sentence sequences containing individual words.",
                        items=types.Schema(
                            type=types.Type.OBJECT,
                            properties={
                                "words": types.Schema(
                                    type=types.Type.ARRAY,
                                    description="The chronological list of words in this phrase block.",
                                    items=types.Schema(
                                        type=types.Type.OBJECT,
                                        properties={
                                            "word": types.Schema(type=types.Type.STRING, description="The single spoken word"),
                                            "start_time": types.Schema(type=types.Type.STRING, description="Start time in HH:MM:SS,mmm"),
                                            "end_time": types.Schema(type=types.Type.STRING, description="End time in HH:MM:SS,mmm"),
                                        },
                                        required=["word", "start_time", "end_time"]
                                    )
                                )
                            },
                            required=["words"]
                        ),
                    ),
                },
                required=["sequences"],
            ),
        ),
    )

    # 5. Parse the JSON response
    try:
        raw_json = response.text.strip("```json\n").strip("```")
        data = json.loads(raw_json)
        sequences = data.get("sequences", [])
    except json.JSONDecodeError as e:
        print(f"Error parsing Gemini's output: {e}")
        print("Raw output:", response.text)
        return

    # 6. Build the ASS Style Headers
    # Style Config: Font Arial Black, Size 80, Centered (Alignment 5), MarginV 200, Thick Black Border
    ass_content = """[Script Info]
ScriptType: v4.00+
PlayResX: 1080
PlayResY: 1920

[V4+ Styles]
Format: Name, Fontname, Fontsize, PrimaryColour, SecondaryColour, OutlineColour, BackColour, Bold, Italic, Underline, StrikeOut, ScaleX, ScaleY, Spacing, Angle, BorderStyle, Outline, Shadow, Alignment, MarginL, MarginR, MarginV, Encoding
Style: HighlightStyle,Arial Black,80,&H00FFFFFF,&H000000FF,&H00000000,&H00000000,-1,0,0,0,100,100,0,0,1,5,0,5,10,10,200,1

[Events]
Format: Layer, Start, End, Style, Name, MarginL, MarginR, MarginV, Effect, Text
"""

    # 7. Process sequences and generate animated frames
    print(f"Compiling animated subtitles to {output_ass_path}...")

    for seq in sequences:
        words_list = seq.get("words", [])
        if not words_list:
            continue

        # For every word in the sentence, we create a dialogue line entry
        # where ONLY that word is color-swapped to yellow
        for i, target_word_data in enumerate(words_list):
            start_ass = srt_to_ass_time(target_word_data.get("start_time", "00:00:00,000"))
            end_ass = srt_to_ass_time(target_word_data.get("end_time", "00:00:00,000"))

            line_text = ""
            for j, w_data in enumerate(words_list):
                word_str = w_data.get("word", "").strip()
                if j == i:
                    # Apply yellow color tag (\1c&H00FFFF&) and revert back to white (\1c&HFFFFFF&)
                    line_text += r"{\1c&H00FFFF&}" + word_str + r"{\1c&HFFFFFF&} "
                else:
                    line_text += word_str + " "

            # Format according to ASS script event layout
            ass_content += f"Dialogue: 0,{start_ass},{end_ass},HighlightStyle,,0,0,0,,{line_text.strip()}\n"

    # Save to file
    with open(output_ass_path, "w", encoding="utf-8") as ass_file:
        ass_file.write(ass_content)

    print("ASS generation complete! Subtitles are ready to be burned into your video.")

if __name__ == "__main__":
    # --- Execute Script ---
    INPUT_AUDIO = "audio.wav"
    OUTPUT_FILE = "transcript.ass"

    generate_ass_from_audio(INPUT_AUDIO, OUTPUT_FILE)
````

#### Caption Generator Node

````python
import subprocess
import os
import re

def parse_time_to_ms(time_str):
    """Converts SRT time 'HH:MM:SS,mmm' to milliseconds."""
    h, m, s_ms = time_str.strip().split(':')
    s, ms = s_ms.split(',')
    return int(h) * 3600000 + int(m) * 60000 + int(s) * 1000 + int(ms)

def ms_to_ass_time(ms):
    """Converts milliseconds to ASS time format 'H:MM:SS.cs'."""
    h = ms // 3600000
    ms = ms % 3600000
    m = ms // 60000
    ms = ms % 60000
    s = ms // 1000
    cs = (ms % 1000) // 10  # Centiseconds (2 digits)
    return f"{h}:{m:02d}:{s:02d}.{cs:02d}"

def srt_to_animated_ass(srt_filename, ass_filename):
    """Parses an SRT file and generates an ASS file with word-by-word highlighting."""
    if not os.path.exists(srt_filename):
        print(f"[-] Error: SRT file '{srt_filename}' not found.")
        return

    # 1. ASS Header with Styling
    ass_content = """[Script Info]
ScriptType: v4.00+
PlayResX: 1080
PlayResY: 1920

[V4+ Styles]
Format: Name, Fontname, Fontsize, PrimaryColour, SecondaryColour, OutlineColour, BackColour, Bold, Italic, Underline, StrikeOut, ScaleX, ScaleY, Spacing, Angle, BorderStyle, Outline, Shadow, Alignment, MarginL, MarginR, MarginV, Encoding
Style: HighlightStyle,Arial Black,80,&H00FFFFFF,&H000000FF,&H00000000,&H00000000,-1,0,0,0,100,100,0,0,1,5,0,5,10,10,200,1

[Events]
Format: Layer, Start, End, Style, Name, MarginL, MarginR, MarginV, Effect, Text
"""

    # 2. Parse the SRT file
    with open(srt_filename, 'r', encoding='utf-8') as f:
        content = f.read()

    # Split by double newlines to get individual subtitle blocks
    blocks = re.split(r'\n\n+', content.strip())

    for block in blocks:
        lines = block.split('\n')
        if len(lines) >= 3:
            # Extract timings
            time_line = lines[1]
            if ' --> ' not in time_line:
                continue
            start_str, end_str = time_line.split(' --> ')
            start_ms = parse_time_to_ms(start_str)
            end_ms = parse_time_to_ms(end_str)

            # Extract text and clean it
            text = " ".join(lines[2:])
            text = re.sub(r'<[^>]+>', '', text) # Remove HTML tags if any exist
            words = text.split()

            if not words:
                continue

            # 3. Approximate word timings (Even distribution)
            duration = end_ms - start_ms
            ms_per_word = duration // len(words)

            # 4. Generate dialogue lines for each highlighted word
            for i, target_word in enumerate(words):
                word_start = start_ms + (i * ms_per_word)
                word_end = word_start + ms_per_word

                line_text = ""
                for j, w in enumerate(words):
                    if j == i:
                        # Highlight current word in Yellow
                        line_text += r"{\1c&H00FFFF&}" + w + r"{\1c&HFFFFFF&} "
                    else:
                        line_text += w + " "

                ass_content += f"Dialogue: 0,{ms_to_ass_time(word_start)},{ms_to_ass_time(word_end)},HighlightStyle,,0,0,0,,{line_text.strip()}\n"

    # Write the ASS file
    with open(ass_filename, "w", encoding="utf-8") as f:
        f.write(ass_content)
    print(f"[+] Subtitle file '{ass_filename}' generated successfully.")

def add_subtitles_to_video(input_video, subtitle_file, output_video):
    """Uses FFmpeg to burn the ASS subtitles directly onto the video stream."""
    if not os.path.exists(input_video):
        print(f"[-] Error: Input video '{input_video}' not found.")
        return

    command = [
        'ffmpeg', '-y',
        '-i', input_video,
        '-vf', f"ass={subtitle_file}",
        '-c:a', 'copy',
        output_video
    ]

    print(f"[+] Running FFmpeg command to burn subtitles...")
    try:
        subprocess.run(command, check=True)
        print(f"[+] Success! Video saved to '{output_video}'")
    except subprocess.CalledProcessError as e:
        print(f"[-] FFmpeg failed with error: {e}")

if __name__ == "__main__":
    # --- Configuration ---         # Replace with your .srt file name
    INPUT_VIDEO = "SlideshowAds.mp4"        # Replace with your input video name
    OUTPUT_VIDEO = "captionvideo.mp4"        # The final video file
    TEMP_ASS_FILE = "transcript.ass"

    # 1. Convert SRT to the animated ASS format
    #srt_to_animated_ass(INPUT_SRT, TEMP_ASS_FILE)

    # 2. Burn into video (Uncomment when you have your files ready)
    add_subtitles_to_video(INPUT_VIDEO, TEMP_ASS_FILE, OUTPUT_VIDEO)
````

#### Background Music Generator Node

````python
%%writefile BackgroundMusicGeneratorNode.py

import os
from moviepy.editor import VideoFileClip, AudioFileClip, CompositeAudioClip
import moviepy.audio.fx.all as afx

def add_background_music(
    video_path,
    music_path,
    output_path="final_video_with_music.mp4",
    video_volume=1.0,
    music_volume=0.1
):
    """
    Adds background music to a video with adjustable volume levels.

    :param video_path: Path to the input video.
    :param music_path: Path to the input background music.
    :param output_path: Path to save the final video.
    :param video_volume: Volume multiplier for the original video (1.0 = 100%, 0.5 = 50%, 0.0 = mute).
    :param music_volume: Volume multiplier for the background music.
    """

    if not os.path.exists(video_path):
        raise FileNotFoundError(f"Video file not found: {video_path}")
    if not os.path.exists(music_path):
        raise FileNotFoundError(f"Music file not found: {music_path}")

    print("Loading video and audio clips...")
    video = VideoFileClip(video_path)

    # 1. Handle Original Video Audio
    if video.audio is not None:
        # Apply volume adjustment to original audio
        original_audio = video.audio.volumex(video_volume)
    else:
        original_audio = None

    # 2. Handle Background Music
    # Load and apply volume adjustment
    bg_music = AudioFileClip(music_path).volumex(music_volume)

    # Sync durations so the music perfectly matches the video length
    if bg_music.duration < video.duration:
        print("Music is shorter than video. Looping audio...")
        bg_music = afx.audio_loop(bg_music, duration=video.duration)
    else:
        print("Music is longer than video. Trimming audio...")
        bg_music = bg_music.set_duration(video.duration)

    # 3. Combine Audio Tracks
    if original_audio is not None and video_volume > 0.0:
        print("Mixing original audio with background music...")
        # Layers the two audio tracks on top of each other
        final_audio = CompositeAudioClip([original_audio, bg_music])
    else:
        print("Replacing original audio entirely...")
        final_audio = bg_music

    # Attach the mixed audio back to the video
    final_video = video.set_audio(final_audio)

    # 4. Export Final Video
    print(f"Exporting final video to {output_path}...")
    final_video.write_videofile(
        output_path,
        codec="libx264",
        audio_codec="aac",
        # Lower threads if it consumes too much CPU, increase for faster rendering
        threads=4,
        preset="medium"
    )

    # 5. Clean up memory
    video.close()
    bg_music.close()
    if original_audio:
        original_audio.close()
    final_audio.close()
    print("Done!")

if __name__ == "__main__":
    # --- Configuration ---
    MY_VIDEO = "captionvideo.mp4"
    MY_MUSIC = "/content/sample_data/leda_backgroundmusic1.mp3"
    OUTPUT = "ImilyBelaWomensSummerTop.mp4"

    # Volume Range Guide:
    # 0.0 = Muted
    # 0.5 = 50% Volume (Quieter)
    # 1.0 = 100% Volume (Normal)
    # 1.5 = 150% Volume (Boosted)

    add_background_music(
        video_path=MY_VIDEO,
        music_path=MY_MUSIC,
        output_path=OUTPUT,
        video_volume=1.5,  # Keep video speaking volume normal
        music_volume=0.05  # Keep background music low (15%) so it doesn't overpower the voice
    )
````

````python
import os
import random
import requests
from urllib.parse import urlparse
from moviepy.editor import ImageClip, AudioFileClip, CompositeVideoClip, ColorClip
import moviepy.video.fx.all as vfx

# --- Configuration ---
# 9:16 Aspect Ratio
TARGET_W = 1080
TARGET_H = 1920
# 16:9 Aspect Ratio#
#TARGET_W = 1920
#TARGET_H = 1080

TRANSITION_DURATION = 1.0  # Duration of transitions in seconds

def process_image_inputs(image_list, download_folder="downloaded_images"):
    """
    Checks the input list for URLs. Downloads URLs to a local folder
    and returns a list of strictly local file paths.
    """
    processed_paths = []

    # Create the download directory if it doesn't exist
    if not os.path.exists(download_folder):
        os.makedirs(download_folder)

    for i, item in enumerate(image_list):
        if item.startswith("http://") or item.startswith("https://"):
            print(f"Downloading image from URL: {item}")
            try:
                response = requests.get(item, stream=True)
                response.raise_for_status() # Check for valid download

                # Try to get the filename from the URL, or generate a fallback name
                parsed_url = urlparse(item)
                filename = os.path.basename(parsed_url.path)
                if not filename or '.' not in filename:
                    filename = f"downloaded_img_{i}.jpg"

                local_path = os.path.join(download_folder, filename)

                # Write the image to the local file
                with open(local_path, 'wb') as f:
                    for chunk in response.iter_content(1024):
                        f.write(chunk)

                processed_paths.append(local_path)
                print(f"Saved to: {local_path}")
            except Exception as e:
                print(f"Error downloading {item}. Skipping. Error: {e}")
        else:
            # It's already a local file path
            if os.path.exists(item):
                processed_paths.append(item)
            else:
                print(f"Warning: Local file not found at {item}. Skipping.")

    return processed_paths

def resize_to_fill(clip, target_w, target_h):
    """Resizes an image to completely cover the target resolution."""
    clip_ratio = clip.w / clip.h
    target_ratio = target_w / target_h

    if clip_ratio > target_ratio:
        return clip.resize(height=target_h)
    else:
        return clip.resize(width=target_w)

def make_animated_clip(img_path, duration, effect):
    """Applies strictly horizontal panning effects (or vertical fallback for tall images)."""
    clip = ImageClip(img_path).set_duration(duration)
    clip = resize_to_fill(clip, TARGET_W, TARGET_H)

    max_x = clip.w - TARGET_W
    max_y = clip.h - TARGET_H

    if max_x > 0:
        # Image is wide enough to pan horizontally
        if effect == "pan_left":
            # Starts right, moves left
            pos_func = lambda t: (-int(max_x * (t / duration)), "center")
        else: # pan_right
            # Starts left, moves right
            pos_func = lambda t: (-int(max_x * (1 - (t / duration))), "center")
    elif max_y > 0:
        # Fallback: If the image is extremely tall and narrow, pan vertically so it isn't static
        if effect == "pan_left": # Treat as pan up
            pos_func = lambda t: ("center", -int(max_y * (t / duration)))
        else: # Treat as pan down
            pos_func = lambda t: ("center", -int(max_y * (1 - (t / duration))))
    else:
        # Perfect fit, no panning needed
        pos_func = lambda t: ("center", "center")

    animated = clip.set_position(pos_func)

    # Base background
    base = ColorClip(size=(TARGET_W, TARGET_H), color=(0, 0, 0)).set_duration(duration)
    return CompositeVideoClip([base, animated], size=(TARGET_W, TARGET_H))

def apply_random_transition(clip, duration):
    """Randomly selects and applies a beautiful blending transition."""
    styles = ["crossfade", "dip_to_black", "flash_to_white"]
    selected_style = random.choice(styles)

    if selected_style == "crossfade":
        return clip.crossfadein(duration)
    elif selected_style == "dip_to_black":
        # Fades in from black
        return clip.fadein(duration)
    elif selected_style == "flash_to_white":
        # Fades in from white, creating a light leak/flash effect
        return clip.fadein(duration, initial_color=[255, 255, 255])

def create_dynamic_slideshow(image_paths, audio_path, output_path="output_slideshow.mp4"):
    num_images = len(image_paths)
    if num_images < 2:
        raise ValueError("You need at least 2 valid images to create a slideshow with transitions.")

    # 1. Load Audio
    audio = AudioFileClip(audio_path)
    total_audio_duration = audio.duration

    # 2. Calculate Timings
    clip_duration = (total_audio_duration + (num_images - 1) * TRANSITION_DURATION) / num_images

    # 3. Create Clips in a Loop
    clips = []

    for i, img_path in enumerate(image_paths):
        # Randomize the panning direction for every image
        effect = random.choice(["pan_left", "pan_right"])
        print(f"Generating Clip {i + 1}/{num_images} ({effect})...")

        clip = make_animated_clip(img_path, clip_duration, effect)

        # Apply start time and a random transition to all clips EXCEPT the first one
        if i > 0:
            start_time = i * (clip_duration - TRANSITION_DURATION)
            clip = clip.set_start(start_time)
            clip = apply_random_transition(clip, TRANSITION_DURATION)

        clips.append(clip)

    # 4. Combine and Export
    print("Compositing final video...")
    video = CompositeVideoClip(clips, size=(TARGET_W, TARGET_H))
    video = video.set_audio(audio).set_duration(total_audio_duration)

    print(f"Writing to {output_path}...")
    video.write_videofile(output_path, fps=30, codec="libx264", audio_codec="aac")

if __name__ == "__main__":
    # You can now mix local files and web URLs in this list
    my_mixed_images = [
        "https://m.media-amazon.com/images/I/71MYEPKFZcL._AC_SY879_.jpg",
        "https://m.media-amazon.com/images/I/81C9vnXzjaL._AC_SY879_.jpg",
        "https://m.media-amazon.com/images/I/71ihQSufepL._AC_SY879_.jpg",
        "https://m.media-amazon.com/images/I/81nlCCCdFzL._AC_SY879_.jpg",
        "https://m.media-amazon.com/images/I/71HMGuwDAKL._AC_SY879_.jpg",
        "https://m.media-amazon.com/images/I/71GtfdKAuBL._AC_SY879_.jpg",
         ]
    my_voiceover = "audio.wav"

    # Pre-process the list to download any URLs
    print("Verifying and preparing images...")
    final_image_paths = process_image_inputs(my_mixed_images)

    # Run the slideshow generator with the processed paths
    create_dynamic_slideshow(final_image_paths, my_voiceover, "SlideshowAds.mp4")
````

#### UGC Video Workflow

````python

import os
import time
import base64
import json
import asyncio
import requests
import google.generativeai as genai
from pathlib import Path
from dotenv import load_dotenv

# For final audio/video muxing
try:
    import moviepy.editor as mp
except ImportError:
    print("Please install moviepy: pip install moviepy")

load_dotenv()

class UGCGenerator:
    def __init__(self):
        # API Keys
        self.google_key = os.getenv("GOOGLE_API_KEY")
        self.kie_key = os.getenv("KIE_API_KEY")

        # Google Config
        genai.configure(api_key=self.google_key)
        self.logic_model = genai.GenerativeModel("gemini-2.5-flash")

        # KIE Config
        self.kie_headers = {"Authorization": f"Bearer {self.kie_key}", "Content-Type": "application/json"}
        self.kie_base_url = "https://api.kie.ai/api/v1"
        self.kie_upload_url = "https://kieai.redpandaai.co/api/file-base64-upload"

    # --- STEP 1 & 2: IMAGE & ASSET PREP ---
    def _upload_to_kie(self, path):
        with open(path, "rb") as f:
            encoded = base64.b64encode(f.read()).decode('utf-8')
        mime = "image/png" if path.endswith(".png") else "image/jpeg"
        res = requests.post(self.kie_upload_url, headers=self.kie_headers,
                            json={"base64Data": f"data:{mime};base64,{encoded}"})
        return res.json()['data']['downloadUrl']

    def generate_hero_scene(self, inf_path, prod_path, details, platform, audience):
        # Gemini 2.5 Flash Direction (Step 1)
        p = Path(inf_path)
        inf_part = {"mime_type": "image/jpeg", "data": p.read_bytes()}
        prompt = f"Create a 4K UGC scene prompt for {platform}. Subject: Image A. Product: {details}. Style: Cinematic, 8K, pores visible."
        brief = self.logic_model.generate_content([prompt, inf_part]).text.strip()

        # Nano Banana Pro Rendering (Step 2)
        inf_url = self._upload_to_kie(inf_path)
        prod_url = self._upload_to_kie(prod_path)
        payload = {
            "model": "nano-banana-pro",
            "input": {"prompt": brief, "image_input": [inf_url, prod_url], "aspect_ratio": "9:16", "resolution": "2K"}
        }
        task = requests.post(f"{self.kie_base_url}/jobs/createTask", headers=self.kie_headers, json=payload).json()

        # Polling (Step 2 Feature)
        while True:
            res = requests.get(f"{self.kie_base_url}/jobs/record-info", headers=self.kie_headers, params={"taskId": task['data']['taskId']}).json()
            if res.get('data', {}).get('status') == "SUCCESS":
                return res['data']['response']['resultUrls'][0], brief, inf_url
            time.sleep(15)

    # --- STEP 3 & 4: VIDEO & EXTENSION ---
    def generate_full_video(self, hero_url, product_name, features, brief):
        # Scripting (Step 3)
        script_prompt = f"Generate a JSON list of 3 scenes (8s each). Context: {brief}. Product: {product_name}. Features: {features}."
        script = json.loads(self.logic_model.generate_content(script_prompt).text.replace("```json", "").replace("```", ""))

        # Veo 3.1 Generation & Recursive Extension (Step 4)
        video_chain = []
        for i, scene in enumerate(script):
            endpoint = "/veo/generate" if i == 0 else "/api/v1/veo/extend"
            payload = {
                "model": "fast",
                "prompt": f"{scene['action']}. Dialogue: {scene['dialogue']}",
                "duration": "8s"
            }
            if i == 0:
                payload["input"] = {"image_input": [hero_url], "aspect_ratio": "9:16"}
            else:
                payload["taskId"] = video_chain[-1]['taskId']

            task = requests.post(f"{self.kie_base_url}{endpoint}", headers=self.kie_headers, json=payload).json()

            # Polling (Step 4 Feature)
            while True:
                res = requests.get(f"{self.kie_base_url}/veo/record-info", headers=self.kie_headers, params={"taskId": task['data']['taskId']}).json()
                if res.get('data', {}).get('status') == "SUCCESS":
                    video_chain.append({"taskId": task['data']['taskId'], "url": res['data']['response']['videoUrl']})
                    break
                time.sleep(30)

        return video_chain[-1]['url'] # Return the final extended video

    # --- STEP 5: VOICE TRANSFORMATION (LEDA) ---
    async def apply_leda_voice(self, video_url):
        """Connects to Gemini Live API to swap voice to Leda."""
        print("[Live API] Initializing Voice Transformation to 'Leda'...")

        # 1. Download final video
        video_data = requests.get(video_url).content
        with open("raw_video.mp4", "wb") as f: f.write(video_data)

        # 2. Extract original audio
        video_clip = mp.VideoFileClip("raw_video.mp4")
        video_clip.audio.write_audiofile("temp_audio.wav")

        # 3. Gemini 3.1 Flash Live Session
        # Note: This is a conceptual async block for the WebSocket Voice Swap
        async with genai.live.connect(model="gemini-3.1-flash-live-preview") as session:
            # Set voice configuration to Leda (Step 14 requirement)
            await session.send(input={
                "setup": {
                    "speech_config": {"voice_config": {"prebuilt_voice_config": {"voice_name": "Leda"}}},
                    "system_instruction": "Listen to the audio and repeat it exactly in your voice."
                }
            }, end_of_turn=True)

            # Stream the temp_audio.wav into the session here...
            # Collect returned Leda PCM bytes...
            print("[Live API] Audio transformation complete.")

        # 4. Muxing (Merging Leda Audio back to Video)
        # Assuming leda_audio.wav was saved from the stream
        # final_video = video_clip.set_audio(mp.AudioFileClip("leda_audio.wav"))
        # final_video.write_videofile("UGC_FINAL_LEDA.mp4")
        return "UGC_FINAL_LEDA.mp4"

# --- MAIN RUN ---
async def main():
    engine = UGCGenerator()
    # Execute Phase 1-4
    hero_url, brief, inf_ref = engine.generate_hero_scene("me.jpg", "product.png", "Glow Serum", "TikTok", "Gen Z")
    final_veo_url = engine.generate_full_video(hero_url, "Glow Serum", "Vitamin C", brief)

    # Execute Phase 5 (Voice Swap)
    final_path = await engine.apply_leda_voice(final_veo_url)
    print(f"Workflow Complete: {final_path}")

if __name__ == "__main__":
    asyncio.run(main())
````
